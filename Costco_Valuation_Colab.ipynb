### 5. Query the run history (SQL)

```python
import sqlite3, pandas as pd

con = sqlite3.connect("valuation.db")
pd.read_sql("""
    SELECT run_timestamp_utc, data_as_of, price, wacc, dcf_value,
           target_price, rating, mc_dcf_p50, mc_share_below_price
    FROM runs ORDER BY run_id DESC
""", con)
```
